#### **Problem #1.4**

### Explain how you can get the integer result 5 from 100 divide by 20

If using normal division with just one / , then the result is a decimal value. However, when two // are used then the result is a floored quotient. This means that when the quotient is a positive value, any fraction that would be part of the answer using normal division is removed. This leaves just the whole number of the quotient. When the quotient is negative this means that the answer produced is rounded down by 1 becoming more negative.
